```python
import pandas as pd
from pathlib import Path

data_folder = Path("DataSets")

field_service = pd.read_excel(
    data_folder / "Competition_Dataset1_Field_Service_Final.xlsx"
)

field_service.head()
```

```python
print("Rows and columns:", field_service.shape)
print("\nColumn names:")
print(field_service.columns.tolist())

print("\nData types:")
print(field_service.dtypes)

print("\nMissing values per column:")
print(field_service.isna().sum())

print("\nCompletely duplicated rows:")
print(field_service.duplicated().sum())
```

```python
# Show each category and how often it occurs.
columns = [
    "Product Family",
    "Customer Region",
    "Failure Category",
    "Warranty Eligible"
]

for column in columns:
    print(f"\n{column}:")
    print(field_service[column].value_counts(dropna=False).to_string())

# Inspect dates before deciding how to convert them.
print("\nExample dates:")
print(field_service["Date"].head(15).to_string(index=False))

# Check whether the same unit appears multiple times.
print("\nRepeated serial numbers beyond their first occurrence:")
print(field_service["Serial Number"].duplicated().sum())

# Check the range of numeric values.
print("\nNumeric summary:")
print(field_service[["Downtime Hours", "Repair Cost"]].describe())
```

```python
print("Example dates:")
print(field_service["Date"].head(10).to_string(index=False))

print("\nRepeated serial numbers beyond the first:")
print(field_service["Serial Number"].duplicated().sum())

print("\nMinimum and maximum numeric values:")
display(
    field_service[["Downtime Hours", "Repair Cost"]]
    .agg(["min", "max"])
)
```
