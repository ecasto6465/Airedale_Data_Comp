```python
#imports
import pandas as pd

#Reading in data sets
field_serviceDF = pd.read_excel('DataSets/Competition_Dataset1_Field_Service_Final.xlsx')
warrantyDF = pd.read_excel('DataSets/Competition_Dataset2_Warranty_Final.xlsx')
productDF = pd.read_excel('DataSets/Competition_Dataset3_Product_Config_Final.xlsx')
reworkDF = pd.read_excel('DataSets/Competition_Dataset4_Rework_Final.xlsx')
eoltDF = pd.read_excel('DataSets/Competition_Dataset5_EOLT_Final.xlsx')

#joining data sets
finalDF = pd.merge(field_serviceDF, warrantyDF, how='left', on='Serial Number', validate = "one_to_one")
finalDF = pd.merge(finalDF, productDF, how='left', on='Serial Number', validate = "one_to_one")
finalDF = pd.merge(finalDF, reworkDF, how='left', on='Serial Number', validate = "one_to_one")
finalDF = pd.merge(finalDF, eoltDF, how='left', on='Serial Number', validate = "one_to_one")

#finalDF = finalDF.rename(columns={'Product family_x': 'Product family', 'Failure Category_x': 'Failure Category})
#finalDF.to_excel('DataSets/Competition_Dataset_Final.xlsx', index=False)
finalDF.head()
```

```python
if(finalDF["Product Family_x"].equals(finalDF["Product Family_y"]) and 
   finalDF["Failure Category_x"].equals(finalDF["Failure Category_y"])
   and finalDF["Facility_x"].equals(finalDF["Facility_y"]) and finalDF["Shift_x"].equals(finalDF["Shift_y"])):
    print("The columns are equal.")

finalDF = finalDF.drop(columns=['Product Family_y', 'Failure Category_y', 'Facility_y', 'Shift_y', "Warranty Eligible", "Pass After Repair", "Retested"])
finalDF = finalDF.rename(columns={'Product Family_x': 'Product Family', 'Failure Category_x': 'Failure Category', 'Facility_x': 'Facility', 'Shift_x': 'Shift', 'Failure Mode' : 'Failure Mode ELOT',
                                  'Failure Mode' : 'Failure Mode Rework', 'Material Cost' : 'Material Cost Rework', 'Repair Cost' : 'Repair Cost Field Service', 'Labor Hours' : 'Labor Hours Rework', 'Date' : 'Service Date'})
finalDF = finalDF.loc[:, ~finalDF.columns.duplicated()]
finalDF['Service Date'] = pd.to_datetime(finalDF['Service Date'])
finalDF['Claim Date'] = pd.to_datetime(finalDF['Claim Date'])
finalDF.head()
finalDF.to_excel('DataSets/Competition_Dataset_Final.xlsx', index=False)
```
